```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('heart_failure_clinical_records_dataset.csv')
df.head()
```

```python
df.tail()
```

```python
df.info()
```

```python
df.isnull().sum()
```

**Separating X and Y**

```python
X = df.drop('DEATH_EVENT', axis=1).values
Y = df['DEATH_EVENT'].values.reshape(-1, 1)
m = len(Y)
```

**Feature Scaling**

```python
X = (X-np.mean(X, axis=0))/np.std(X, axis=0)
```

**Add Bias Column**

```python
X = np.c_[np.ones((m,1)), X]
```

**Applying Multiple Linear Regression**

```python
Xt = X.T
beta = np.linalg.inv(Xt @ X) @ Xt @ Y

print("Beta coefficients:\n", beta)

```

**Output Z**

```python
z =  X @ beta
print("z: ",z)
```

**Finding Sigmoid**

```python
def sigmoid(z):
  return 1/(1+np.exp(-z))

y_pred = sigmoid(z)
print("y_pred: ",y_pred)
```

**Class Prediction**

```python
y_pred = (y_pred >= 0.5).astype(int)
print("y_pred: ",y_pred)
```

**Accuracy**

```python
accuracy = np.mean(y_pred == Y) * 100
print("Model Accuracy =", accuracy, "%")
```

**Plot Sigmoid**

```python
plt.scatter(z, y_pred)
plt.xlabel("Linear Output z")
plt.ylabel("Sigmoid Probability")
plt.title("Logistic Curve")
plt.show()
```
